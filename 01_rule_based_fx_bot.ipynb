```python
import os
from getpass import getpass
import requests
import pandas as pd

if not os.environ.get("EXCHANGE_API_KEY"):
    os.environ["EXCHANGE_API_KEY"] = getpass("Paste your ExchangeRate-API key: ")

api_key = os.environ["EXCHANGE_API_KEY"]
url = f"https://v6.exchangerate-api.com/v6/{api_key}/latest/USD"

response = requests.get(url, timeout=10)
data = response.json()

if response.status_code == 200 and data["result"] == "success":
    rates = data["conversion_rates"]
    selected = ["PKR", "EUR", "GBP", "AED", "SAR", "CNY", "JPY"]
    df_rates = pd.DataFrame([
        {
            "currency": code,
            "1_usd_equals": rates[code],
            "100_usd_equals": round(100 * rates[code], 2),
            "last_updated": data["time_last_update_utc"]
        }
        for code in selected
    ])
    print("Connected! Live rates loaded:")
    display(df_rates)
else:
    print("API Error:", data)
```

```python
import os
from getpass import getpass
import requests

def currency_chatbot():
    if not os.environ.get("EXCHANGE_API_KEY"):
        os.environ["EXCHANGE_API_KEY"] = getpass("Paste your ExchangeRate-API key: ")
        
    api_key = os.environ["EXCHANGE_API_KEY"]
    print("💱 Currency Bot Ready! Type like: '100 USD to PKR' (or 'quit' to exit)\n")
    
    while True:
        user_msg = input("You: ").strip()
        
        if user_msg.lower() in ["quit", "exit", "bye"]:
            print("Bot: Goodbye! Have a great day.")
            break
            
        try:
            parts = [word for word in user_msg.upper().split() if word != "TO"]
            amount = float(parts[0])
            from_curr = parts[1]
            to_curr = parts[2]
            
            url = f"https://v6.exchangerate-api.com/v6/{api_key}/pair/{from_curr}/{to_curr}/{amount}"
            res = requests.get(url, timeout=10).json()
            
            if res.get("result") == "success":
                converted = res["conversion_result"]
                rate = res["conversion_rate"]
                print(f"Bot: {amount:,.2f} {from_curr} = {converted:,.2f} {to_curr} (Rate: 1 {from_curr} = {rate} {to_curr})\n")
            else:
                print(f"Bot: Couldn't convert that ({res.get('error-type', 'check currency codes')}).\n")
                
        except (ValueError, IndexError):
            print("Bot: Please use the format: 100 USD to PKR\n")
        except requests.RequestException as e:
            print(f"Bot: Network error ({e}). Please check your internet connection.\n")

currency_chatbot()
```
